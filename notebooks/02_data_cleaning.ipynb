```python
import pandas as pd
import numpy as np

matches = pd.read_csv('../data/raw/matches.csv')
deliveries = pd.read_csv('../data/raw/deliveries.csv')
print("Loaded:", matches.shape, deliveries.shape)
```

```python
matches = matches.dropna(subset=['winner'])
print("After dropping nulls:", matches.shape)
```

```python
season_map = {'2007/08': '2008', '2009/10': '2010', '2020/21': '2020'}
matches['season'] = matches['season'].replace(season_map).astype(int)
print("Seasons:", sorted(matches['season'].unique()))
```

```python
team_name_map = {
    'Delhi Daredevils'            : 'Delhi Capitals',
    'Kings XI Punjab'             : 'Punjab Kings',
    'Royal Challengers Bangalore' : 'Royal Challengers Bengaluru',
    'Rising Pune Supergiant'      : 'Rising Pune Supergiants',
}
for col in ['team1', 'team2', 'toss_winner', 'winner']:
    matches[col] = matches[col].replace(team_name_map)
for col in ['batting_team', 'bowling_team']:
    deliveries[col] = deliveries[col].replace(team_name_map)

print("Teams:", sorted(matches['team1'].unique()))
print("Total teams:", matches['team1'].nunique())
```

```python
matches['toss_won_match'] = (matches['toss_winner'] == matches['winner']).astype(int)
print("Toss win %:", round(matches['toss_won_match'].mean() * 100, 2), "%")
```

```python
matches.to_csv('../data/processed/matches_clean.csv', index=False)
deliveries.to_csv('../data/processed/deliveries_clean.csv', index=False)
print("Saved successfully!")
print("Matches:", matches.shape)
print("Deliveries:", deliveries.shape)
```
